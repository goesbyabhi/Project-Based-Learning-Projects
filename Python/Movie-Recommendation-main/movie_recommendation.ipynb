# Building a movie recommendation engine using Python

##Importing the libraries

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import CountVectorizer
```

##Loading the data

```python
from google.colab import files
uploaded = files.upload()
```

##Storing the data

```python
df = pd.read_csv('IMDB-Movie-Data.csv')
df['Movie_id']=range(0,1000)
df.head(5)
```

##Getting the count of number of row/movies in the data and the number of columns

```python
df.shape
```

##Creating a list of important columns for the recommendation engline

```python
columns = ['Actors', 'Director','Genre', 'Title']
```

##Showing the data

```python
df[columns].head(5)
```

##Checking for missing values

```python
df[columns].isnull().values.any()
```

Creating a function to combine values of important columns in one string

```python
def get_important_features(data):
  important_features = []
  for i in range(0, data.shape[0]):
    important_features.append(data['Actors'][i]+' '+data['Director'][i]+' '+data['Genre'][i]+' '+data['Title'][i])

  return important_features
```

##Create a column to hold the combined strings

```python
df['important_features'] = get_important_features(df)

df.head(5)
```

##Convert the text into a matrix of token counts

```python
cm = CountVectorizer().fit_transform(df['important_features'])
```

##Get cosine similarity matrix from the count matric

```python
cs = cosine_similarity(cm)

print(cs)
```

##Get the shape of the cosine similarity matrix

```python
cs.shape
```

##Get the title of the movie that the user likes

```python
title = 'The Amazing Spider-Man'
```

##Finding the movies_id

```python
movie_id = df[df.Title==title]['Movie_id'].values[0]
```

##Create a list of enumerations for the similarity score

```python
scores = list(enumerate(cs[movie_id]))
```

#Sorting the list

```python
sorted_scores = sorted(scores, key = lambda x:x[1], reverse = True)
sorted_scores = sorted_scores[1:]
```

##Printing the sorted scores (not necessary so this part is commented out

```python
#print(sorted_scores)
```

##Creating a loop to print the first 7 similar movies

```python
j = 0
print('The 7 most recommended movies to ', title, ' are:\n')
for item in sorted_scores:
  movie_title = df[df.Movie_id== item[0]]['Title'].values[0]
  print(j+1, movie_title)
  j=j+1
  if j>6:
    break
```
