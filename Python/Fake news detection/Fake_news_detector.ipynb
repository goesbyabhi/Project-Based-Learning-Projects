```python
import numpy as np
import pandas as pd
import itertools
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
from google.colab import drive
drive.mount("/content/gdrive")
```

```python
df = pd.read_csv('/content/gdrive/MyDrive/Fake News/news.csv')

df.shape
df.head(10)
```

```python
labels = df.label
labels.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(df['text'], labels, test_size=0.2, random_state=7)
```

```python
tfid_vectorizer = TfidfVectorizer(stop_words="english", max_df=0.7)

tfid_train = tfid_vectorizer.fit_transform(x_train)
tfid_test = tfid_vectorizer.transform(x_test)
```

```python
pac = PassiveAggressiveClassifier(max_iter=50)
pac.fit(tfid_train, y_train)

y_pred = pac.predict(tfid_test)
score = accuracy_score(y_test, y_pred)
print(f'Accuracy: {round(score*100, 2)}%')
```

```python
confusion_matrix(y_test, y_pred, labels=['FAKE', 'REAL'])
```

So with this model, we have 592 true positives, 587 true negatives, 42 false positives, and 46 false negatives.
